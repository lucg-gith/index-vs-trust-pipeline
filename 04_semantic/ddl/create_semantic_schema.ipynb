# Semantic — schema

Thin views over Gold for the dashboard. **No tables, no new arithmetic.** Every number here
already exists in `gold.fact_horizon_performance` or `gold.fact_monthly_performance`; these
views only join, label and group.

That rule is deliberate and it is the answer to *"where does this number come from?"* —
every figure on the dashboard traces to a Gold table, and none of it is computed in the BI
layer where nobody can review or test it.

```sql
CREATE SCHEMA IF NOT EXISTS `index-vs-trust-pipeline`.semantic
COMMENT 'Thin views over Gold for the dashboard. No tables, no arithmetic';
```
