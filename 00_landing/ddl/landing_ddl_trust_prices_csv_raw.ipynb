# `landing.trust_prices_csv_raw` — DDL

The archived monthly price CSV, all 16,357 rows, landed in full even though Silver reads only BCPT and CSH from it.

**The column list is a snapshot of what the source returned on 2026-09-21, not a shape we
chose.** Declaring it means a new source column makes the load fail rather than land, which
is the cost of having Landing declare anything at all. See
`specs/05_orchestration/ddl-split.md`.

```sql
CREATE TABLE IF NOT EXISTS `index-vs-trust-pipeline`.landing.trust_prices_csv_raw (
  trust_name        STRING,
  ticker            STRING,
  aic_sector        STRING,
  date              STRING,
  price_gbx_or_gbp  STRING
)
COMMENT 'Archived monthly trust prices, untransformed';
```
