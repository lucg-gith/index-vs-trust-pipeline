# `landing.trust_prices_yf_raw` — DDL

Monthly bars for the 118 trust tickers, as yfinance returned them.

**The column list is a snapshot of what the source returned on 2026-09-21, not a shape we
chose.** Declaring it means a new source column makes the load fail rather than land, which
is the cost of having Landing declare anything at all. See
`specs/05_orchestration/ddl-split.md`.

```sql
CREATE TABLE IF NOT EXISTS `index-vs-trust-pipeline`.landing.trust_prices_yf_raw (
  source_ticker   STRING,
  symbol          STRING,
  `Date`          TIMESTAMP,
  `Open`          DOUBLE,
  `High`          DOUBLE,
  `Low`           DOUBLE,
  `Close`         DOUBLE,
  `Adj_Close`     DOUBLE,
  `Volume`        BIGINT,
  `Dividends`     DOUBLE,
  `Stock_Splits`  DOUBLE
)
COMMENT 'Yahoo monthly trust bars, untransformed';
```

Expected: 11 columns.
