# Landing — schema

Receiving zone. Data stays exactly as the source sent it.

Catalog name has hyphens, so every reference needs backticks.

```sql
CREATE SCHEMA IF NOT EXISTS `index-vs-trust-pipeline`.landing
COMMENT 'Sources exactly as they arrived';
```

## Pull log

Every other Landing table takes its shape from its source, so the schema is inferred.
This one is our own record of what Yahoo answered per symbol, so it is declared here.

It exists because a symbol Yahoo refuses returns no rows, and would therefore leave no
trace anywhere in the warehouse. Sixteen trust tickers are in that position, and they are
the evidence behind the survivorship-bias argument.

Both ingest notebooks write here. Each replaces only its own `asset_class` slice, so
either can be re-run alone.

```sql
CREATE TABLE IF NOT EXISTS `index-vs-trust-pipeline`.landing.yf_pull_log_raw (
  asset_class      STRING    COMMENT 'trust or index',
  source_ticker    STRING    COMMENT 'ticker as the metadata CSV spells it',
  requested_symbol STRING    COMMENT 'symbol sent to Yahoo, .L suffixed for LSE listings',
  status           STRING    COMMENT 'OK, NODATA or ERROR',
  row_count        INT       COMMENT 'bars returned, 0 when Yahoo does not know the symbol',
  currency         STRING    COMMENT 'GBp, USD or EUR -- Silver cannot rescale what it cannot see',
  first_bar        STRING,
  last_bar         STRING,
  message          STRING    COMMENT 'error text when status is NODATA or ERROR',
  pulled_at        TIMESTAMP
)
COMMENT 'One row per symbol requested from Yahoo, including symbols that returned nothing';
```
