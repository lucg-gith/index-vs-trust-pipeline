# `bronze.yf_pull_log` — DDL

The pull log recast to STRING. Silver reads pulled_at from here to find the partial month.

Declared here rather than inside the load, so the shape can be read without reading the
code that fills it.

```sql
CREATE TABLE IF NOT EXISTS `index-vs-trust-pipeline`.bronze.yf_pull_log (
  asset_class       STRING,
  source_ticker     STRING,
  requested_symbol  STRING,
  status            STRING,
  row_count         STRING,
  currency          STRING,
  first_bar         STRING,
  last_bar          STRING,
  message           STRING,
  pulled_at         STRING
)
COMMENT 'landing.yf_pull_log_raw recast to STRING';
```

## Verification

Expected: 10 columns.
