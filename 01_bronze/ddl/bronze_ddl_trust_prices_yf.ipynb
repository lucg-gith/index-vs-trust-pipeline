# `bronze.trust_prices_yf` — DDL

The Yahoo trust bars recast to STRING. Typing is Silver's job, not Bronze's.

Declared here rather than inside the load, so the shape can be read without reading the
code that fills it.

```sql
CREATE TABLE IF NOT EXISTS `index-vs-trust-pipeline`.bronze.trust_prices_yf (
  source_ticker   STRING,
  symbol          STRING,
  `Date`          STRING,
  `Open`          STRING,
  `High`          STRING,
  `Low`           STRING,
  `Close`         STRING,
  `Adj_Close`     STRING,
  `Volume`        STRING,
  `Dividends`     STRING,
  `Stock_Splits`  STRING
)
COMMENT 'landing.trust_prices_yf_raw recast to STRING';
```
