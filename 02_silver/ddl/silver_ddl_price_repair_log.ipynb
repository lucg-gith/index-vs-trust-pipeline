# `silver.price_repair_log` — DDL

One row per price Silver changed or removed, with the evidence for the decision. The artifact that makes the cleaning defensible.

Declared here rather than inside the load, so the shape can be read without reading the
code that fills it.

```sql
CREATE TABLE IF NOT EXISTS `index-vs-trust-pipeline`.silver.price_repair_log (
  ticker                 STRING,
  month_key              INT,
  month_start            DATE,
  price_source           STRING,
  close_before           DOUBLE,
  close_after            DOUBLE,
  high                   DOUBLE,
  low                    DOUBLE,
  neighbour_median       DOUBLE,
  ratio_to_neighbours    DOUBLE,
  split_factor           DOUBLE,
  close_outside_own_bar  BOOLEAN,
  outcome                STRING,
  reason                 STRING
)
COMMENT 'Every price Silver changed or removed, with the evidence for the decision';
```

Expected: 14 columns.
