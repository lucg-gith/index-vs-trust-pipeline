# `landing.trusts_raw` — DDL

The AIC metadata CSV, exactly as it arrived.

**The column list is a snapshot of what the source returned on 2026-09-21, not a shape we
chose.** Declaring it means a new source column makes the load fail rather than land, which
is the cost of having Landing declare anything at all. See
`specs/05_orchestration/ddl-split.md`.

```sql
CREATE TABLE IF NOT EXISTS `index-vs-trust-pipeline`.landing.trusts_raw (
  trust_name        STRING,
  ticker            STRING,
  aic_sector        STRING,
  source_url        STRING,
  manager           STRING,
  management_group  STRING,
  notes             STRING
)
COMMENT 'uk_investment_trusts.csv, untransformed';
```

Expected: 7 columns.
