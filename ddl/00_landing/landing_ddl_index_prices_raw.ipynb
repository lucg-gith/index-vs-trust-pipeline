# `landing.index_prices_raw` — DDL

Monthly bars for the index trackers. Carries Capital_Gains, which the trust feed does not.

**The column list is a snapshot of what the source returned on 2026-09-21, not a shape we
chose.** Declaring it means a new source column makes the load fail rather than land, which
is the cost of having Landing declare anything at all. See
`specs/05_orchestration/ddl-split.md`.

```sql
CREATE TABLE IF NOT EXISTS `index-vs-trust-pipeline`.landing.index_prices_raw (
  source_ticker    STRING,
  symbol           STRING,
  `Date`           TIMESTAMP,
  `Open`           DOUBLE,
  `High`           DOUBLE,
  `Low`            DOUBLE,
  `Close`          DOUBLE,
  `Adj_Close`      DOUBLE,
  `Volume`         BIGINT,
  `Dividends`      DOUBLE,
  `Stock_Splits`   DOUBLE,
  `Capital_Gains`  DOUBLE
)
COMMENT 'Yahoo monthly index bars, untransformed';
```

## Verification

Expected: 12 columns.

```sql
SELECT COUNT(*) AS columns
FROM `index-vs-trust-pipeline`.information_schema.columns
WHERE table_schema = 'landing' AND table_name = 'index_prices_raw';
```
