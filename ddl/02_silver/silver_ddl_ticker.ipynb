# `silver.ticker` — DDL

One row per ticker, carrying the coverage facts and the status Gold turns into SCD2 versions.

Declared here rather than inside the load, so the shape can be read without reading the
code that fills it. Catalog name has hyphens, so every reference needs backticks.

```sql
CREATE TABLE IF NOT EXISTS `index-vs-trust-pipeline`.silver.ticker (
  ticker            STRING  COMMENT 'Business key',
  trust_name        STRING,
  entity_type       STRING  COMMENT 'Trust or Index, so Gold compares them by self-join',
  aic_sector        STRING,
  manager           STRING  COMMENT 'Full comma-separated list, nulls carried not defaulted',
  management_group  STRING,
  manager_structure STRING  COMMENT 'sole or multi, from the count of named managers',
  currency          STRING,
  price_source      STRING  COMMENT 'yahoo or none',
  source_url        STRING  COMMENT 'Where the trust metadata came from. Cited, not assumed',
  first_month       INT,
  last_month        INT,
  months_available  INT,
  status            STRING  COMMENT 'Always active: the load keeps only trusts still listed',
  data_status       STRING  COMMENT 'usable, stub, excluded or no-data'
)
COMMENT 'Every listed trust and index in the universe, including those with no usable prices';
```
