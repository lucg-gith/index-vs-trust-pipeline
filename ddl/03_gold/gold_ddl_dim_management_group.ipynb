# `gold.dim_management_group` — DDL

The house a trust belongs to. Single-valued, so it needs no bridge: the fact carries
`management_group_key` and joins this table directly.

Declared here rather than inside the load, so the shape can be read without reading the
code that fills it. Catalog name has hyphens, so every reference needs backticks.

```sql
CREATE TABLE IF NOT EXISTS `index-vs-trust-pipeline`.gold.dim_management_group (
  management_group_key STRING COMMENT 'MD5(management_group). The key the fact carries',
  management_group     STRING COMMENT 'NoInfo or NotApplicable, never blank',
  trusts_managed       INT    COMMENT 'Trusts in the house, so a thin dimension still says something'
)
COMMENT 'The houses the trusts belong to, joined straight to the fact';
```
