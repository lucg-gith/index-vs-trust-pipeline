# `gold.dim_mandate` — DDL

Where the money goes, and what it buys. One row per `aic_sector` value.

The sector string does three jobs at once — geography, asset class and style — so on its own
it answers none of them cleanly, and it spreads 102 tickers over 32 buckets. This dimension
splits it into the three columns it was always hiding.

The fact carries `mandate_key` directly, exactly as it carries `management_group_key`, so
the star stays pure.

Declared here rather than inside the load, so the shape can be read without reading the
code that fills it. Catalog name has hyphens, so every reference needs backticks.

```sql
CREATE TABLE IF NOT EXISTS `index-vs-trust-pipeline`.gold.dim_mandate (
  mandate_key       STRING COMMENT 'MD5(aic_sector). The key the fact carries',
  aic_sector        STRING COMMENT 'The published value, unchanged. Natural key of this table',
  region            STRING COMMENT 'Where it invests. NotStated where the AIC classifies by asset instead',
  asset_class       STRING COMMENT 'What it buys: Equity, Private Equity, Infrastructure, Debt, Property, Commodities, Multi-asset',
  style             STRING COMMENT 'Income, Smaller Companies or Broad',
  trusts_in_mandate INT    COMMENT 'How many tickers sit in it'
)
COMMENT 'The 32 AIC sectors, split into the region, asset class and style they encode';
```
