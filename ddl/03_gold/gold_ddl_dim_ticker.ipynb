# `gold.dim_ticker` — DDL

The SCD2 dimension holding trusts and the index together, so trust-versus-index is a self-join on one fact rather than a union.

Versioned on **`manager` and `management_group`**. `status` is not a driver: the table holds
only the trusts that are currently listed, so status can never change within it — a trust that
winds up is removed rather than version-closed. See `specs/08_listed_only/listed-only.md`.

Declared here rather than inside the load, so the shape can be read without reading the
code that fills it. Catalog name has hyphens, so every reference needs backticks.

```sql
CREATE TABLE IF NOT EXISTS `index-vs-trust-pipeline`.gold.dim_ticker (
  ticker_key            STRING  COMMENT 'MD5(ticker|effective_start_month). The key the facts carry',
  ticker                STRING  COMMENT 'Business key. Not unique here: one row per version',
  trust_name            STRING,
  entity_type           STRING  COMMENT 'Trust or Index, so trust-vs-index is a self-join',
  aic_sector            STRING,
  manager               STRING  COMMENT 'SCD2 driver. NoInfo or NotApplicable, never blank',
  management_group      STRING  COMMENT 'SCD2 driver',
  manager_structure     STRING  COMMENT 'sole or multi, from the count of named managers',
  status                STRING  COMMENT 'Always active. Not a driver: a delisted trust leaves the table',
  currency              STRING,
  price_source          STRING  COMMENT 'yahoo or none',
  data_status           STRING  COMMENT 'usable, stub, no-data or excluded',
  source_url            STRING  COMMENT 'Where the metadata came from. Cited, not assumed',
  first_month           INT,
  last_month            INT,
  months_available      INT,
  effective_start_month INT     COMMENT 'YYYYMM this version became true',
  effective_end_month   INT     COMMENT 'YYYYMM it stopped being true. Null while current',
  is_current            BOOLEAN
)
COMMENT 'Every listed trust and the index, versioned on manager and management group';
```
