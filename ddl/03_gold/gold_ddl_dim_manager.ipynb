# `gold.dim_manager` — DDL

One row per named individual, split out of the comma-separated manager list on
`silver.ticker`. Not one row per management team: that would be 97 rows for 97 trusts, a
copy of `dim_ticker` under another name.

Reached from the fact through `bridge_ticker_manager`, because a trust has several managers
and a manager can have several trusts.

Declared here rather than inside the load, so the shape can be read without reading the
code that fills it. Catalog name has hyphens, so every reference needs backticks.

```sql
CREATE TABLE IF NOT EXISTS `index-vs-trust-pipeline`.gold.dim_manager (
  manager_key    STRING COMMENT 'MD5(manager_name). The key the bridge carries',
  manager_name   STRING COMMENT 'One individual, trimmed out of the comma-separated list',
  trusts_managed INT    COMMENT 'One for 224 of them; two for the three who run a pair'
)
COMMENT 'Every named manager in the listed universe, one row each';
```
